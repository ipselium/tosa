# Timed input

```python
from threading import Timer
import time

timeout = 5
t = Timer(timeout, print, ['Sorry, times up'])
t.start()
prompt = "You have %d seconds to choose the correct answer...\n" % timeout
ti = time.perf_counter()
answer = input(prompt)
print('in {:.1f} sec'.format(time.perf_counter() - ti))
t.cancel()
```
